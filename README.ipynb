# Face Detection and Gaze Analysis Pipeline

A comprehensive pipeline for detecting faces in videos, matching them with known actors, and analyzing eye gaze patterns. This project combines computer vision, face recognition, and gaze tracking to create a powerful tool for visual attention analysis.

## Pipeline Overview

1. **Face Detection & Extraction** (Step 1)
   - Detects faces in video frames using a pre-trained Caffe model
   - Extracts and saves detected faces with metadata
   - Processes videos at configurable intervals

2. **Face Matching** (Step 2)
   - Generates embeddings for reference actor images
   - Matches extracted faces with known actors
   - Creates similarity scores for each match

3. **Gaze Integration** (Step 3)
   - Integrates eye tracking data with face detection results
   - Maps gaze points to detected faces
   - Associates viewer attention with recognized actors

4. **Visualization** (Step 4)
   - Creates visual representations of gaze patterns
   - Generates plots showing attention distribution
   - Helps analyze viewing behavior patterns

5. **Model Training** (Step 5)
   - Trains a custom face recognition model
   - Uses matched faces as training data
   - Improves recognition accuracy for specific actors

## Prerequisites

Before running the pipeline, ensure you have the following structure and files:

### Required Directories and Files

1. **Configuration (`config/`)**
   - config_step_1.py through config_step_5.py
   - Update paths in configs to match your system

2. **Data Directories (`data/`)**
   - `actors_ready/`: Reference images of actors (one folder per actor)
   - `videos/`: Input video files (.mp4, .avi, .mov)
   - `gazes_csv/`: Eye tracking data in CSV format
   - `csvs_outputs/`: Empty directory for generated CSVs

3. **Utility Files (`utils/`)**
   - `deploy.prototxt.txt`: Face detection model configuration
   - `res10_300x300_ssd_iter_140000.caffemodel`: Pre-trained face detection model

 #### Actor Image Downloader (`utils/get_actors_ready.py`)
A helpful utility script that can automatically populate your `actors_ready/` directory with actor reference images from IMDb. While not mandatory for the pipeline, it's useful if you don't have a pre-existing collection of actor reference images.

#### Features:
- Downloads actor images from IMDb cast pages
- Automatically creates organized folder structure
- Sanitizes filenames for cross-platform compatibility
- Creates individual folders for each actor/character
- Handles error cases gracefully

#### Usage:
bash
python utils/get_actors_ready.py

When prompted:
1. Enter the IMDb cast URL (e.g., "https://www.imdb.com/title/ttXXXXXXX/fullcredits")
2. Specify how many actors you want to download

The script will create a structured directory:
actors_ready/
└── ActorName_as_CharacterName/
└── ActorName_as_CharacterName.jpg
This utility is particularly helpful when starting a new analysis project and need to quickly set up your actor reference database.

# Step-by-Step Guide

### Step 1: Face Detection and Extraction

write it in your bash/terminal (after you nevigated to the right directory.)

python step_1_face_extraction.py

**Purpose**: Detects and extracts faces from input videos.

**Input**:
- Videos in `data/videos/`
- Face detection model files in `utils/`
  - `deploy.prototxt.txt`
  - `res10_300x300_ssd_iter_140000.caffemodel`

**Output**:
- Extracted faces in `data/faces/{video_name}/`
- Detection metadata in `data/csvs_outputs/{video_name}.csv`

**Process**:
1. Loads pre-trained Caffe model
2. Processes videos frame by frame
3. Extracts detected faces
4. Generates CSV with face locations

### Step 2: Face Matching

python step_2_find_best_match.py

**Purpose**: Matches extracted faces with reference actor images.

**Input**:
- Extracted faces from Step 1 in `data/faces/`
- Actor reference images in `data/actors_ready/`

**Output**:
- Face matching results in `data/csvs_outputs/{video_name}_face_matches.csv`
- Merged location and matching data in `data/csvs_outputs/{video_name}_merged.csv`

**Process**:
1. Generates embeddings for actor reference images
2. Processes extracted faces
3. Finds best matching actor for each face
4. Merges face locations with match results

### Step 3: Gaze Integration


python step_3_integrate_eye_gazes.py

**Purpose**: Integrates eye tracking data with face detection results.

**Input**:
- Eye gaze data in `data/gazes_csv/gazes_{video_name}.csv`
- Merged face data from Step 2

**Output**:
- Integrated results in `data/csvs_outputs/{video_name}_with_gazes.csv`

**Process**:
1. Maps gaze points to video frames
2. Associates gazes with detected faces
3. Links gaze points to recognized actors
4. Generates comprehensive CSV with all data

### Step 4: Visualization


python step_4_visualize.py

**Purpose**: Creates visual representations of the analysis.

**Input**:
- Integrated gaze data from Step 3
- Actor reference images

**Output**:
- Visualization plots in `visualizations/`
  - Gaze distribution patterns
  - Actor attention heatmaps
  - Time-based analysis graphs

**Process**:
1. Loads integrated data
2. Processes temporal patterns
3. Generates visualization plots
4. Saves high-resolution images

### Step 5: Model Training (Optional)


python step_5_train_model.py

**Purpose**: Trains a custom face recognition model using collected data.

**Input**:
- Matched faces from previous steps
- Actor reference images

**Output**:
- Trained model weights in specified output path
- Training metrics and logs

**Process**:
1. Prepares training dataset
2. Initializes model architecture
3. Trains model on matched faces
4. Saves trained model weights

**Note**: This step is optional and can be used to improve face recognition accuracy for specific actors in your dataset.
